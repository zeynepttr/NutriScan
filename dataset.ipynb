```python
import os, shutil, random

final = "/content/final"
train = final + "/train"
val = final + "/val"

os.makedirs(train, exist_ok=True)
os.makedirs(val, exist_ok=True)
```

```python
turkish = "/content/turkish/Turkish Food"

for cls in os.listdir(turkish):
    path = os.path.join(turkish, cls)
    images = os.listdir(path)
    random.shuffle(images)

    split = int(len(images)*0.8)

    for img in images[:split]:
        os.makedirs(f"{train}/{cls}", exist_ok=True)
        shutil.copy(os.path.join(path, img), f"{train}/{cls}")

    for img in images[split:]:
        os.makedirs(f"{val}/{cls}", exist_ok=True)
        shutil.copy(os.path.join(path, img), f"{val}/{cls}")
```

```python
food = "/content/food-101/images"
max_per_class = 200

for cls in os.listdir(food):
    path = os.path.join(food, cls)
    images = os.listdir(path)
    random.shuffle(images)

    split = int(max_per_class * 0.8)

    for img in images[:split]:
        os.makedirs(f"{train}/{cls}", exist_ok=True)
        shutil.copy(os.path.join(path, img), f"{train}/{cls}")

    for img in images[split:max_per_class]:
        os.makedirs(f"{val}/{cls}", exist_ok=True)
        shutil.copy(os.path.join(path, img), f"{val}/{cls}")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8m-cls.pt")

model.train(
    data="/content/final",
    epochs=20,
    imgsz=224,
    batch=32,
    device=0,
    augment=True
)
```
